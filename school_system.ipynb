School Data Integration & Advanced Analysis

#### Objective

The purpose of this analysis is to combine all school datasets into a single dataset for advanced reporting and analytics. This enables comprehensive student profiles, overall performance analysis, attendance analysis, fee analysis, and professional dashboard development.

```python
on=["Student_ID","Roll_No"]
```

Merge Students and Marks


```python
student_marks = pd.merge(
    students,
    marks,
    on="Student_ID",
    how="inner"
)

student_marks.head()
```

Merge Students and Attendance

```python
student_attendance = pd.merge(
    students,
    attendance,
    on="Student_ID",
    how="inner"
)

student_attendance.head()
```

Merge Students and Fees

```python
student_fees = pd.merge(
    students,
    fees,
    on="Student_ID",
    how="inner"
)

student_fees.head()
```

 Remove Duplicate Columns Before Merge

```python
# Marks
marks_merge = marks.drop(columns=["Class", "Academic_Session"])

# Attendance
attendance_merge = attendance.drop(columns=["Class", "Academic_Session"])

# Fees
fees_merge = fees.drop(columns=["Class", "Academic_Session"])
```

Merge All CSV Files

```python
school = (
    students
    .merge(marks_merge, on="Student_ID", how="left")
    .merge(attendance_merge, on="Student_ID", how="left")
    .merge(fees_merge, on="Student_ID", how="left")
)

school.head()
```

```python
import pandas as pd
import numpy as np

# Combine marks and attendance for the same student,
# class, and academic session.
ml_data = pd.merge(
    marks,
    attendance,
    on=["Student_ID", "Academic_Session", "Class"],
    how="inner",
    suffixes=("_marks", "_attendance")
)

# Extract the starting year of each academic session.
ml_data["Session_Year"] = (
    ml_data["Academic_Session"]
    .str.extract(r"(\d{4})")
    .astype(int)
)

# Sort records chronologically for each student.
ml_data = ml_data.sort_values(
    ["Student_ID", "Session_Year"]
).reset_index(drop=True)

# Find the next recorded session for each student.
ml_data["Next_Percentage"] = (
    ml_data.groupby("Student_ID")["Percentage"].shift(-1)
)

ml_data["Next_Session_Year"] = (
    ml_data.groupby("Student_ID")["Session_Year"].shift(-1)
)

# Keep only consecutive academic sessions.
ml_data = ml_data[
    ml_data["Next_Session_Year"]
    == ml_data["Session_Year"] + 1
].copy()

# Remove records without a valid target.
ml_data = ml_data.dropna(subset=["Next_Percentage"])

print("Training records:", len(ml_data))
print("Unique students:", ml_data["Student_ID"].nunique())
print("Input columns:", ml_data.columns.tolist())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error, r2_score
import joblib

# Inputs from the previous session only.
numeric_features = [
    "English",
    "Urdu",
    "Mathematics",
    "Science",
    "Computer",
    "Islamiat",
    "Attendance_Percentage"
]

categorical_features = ["Class"]

feature_columns = numeric_features + categorical_features

X = ml_data[feature_columns]
y = ml_data["Next_Percentage"]

# Use the latest target session as the test set.
latest_target_year = int(ml_data["Next_Session_Year"].max())

train_mask = ml_data["Next_Session_Year"] < latest_target_year
test_mask = ml_data["Next_Session_Year"] == latest_target_year

X_train = X.loc[train_mask]
X_test = X.loc[test_mask]
y_train = y.loc[train_mask]
y_test = y.loc[test_mask]

if len(X_train) == 0 or len(X_test) == 0:
    raise ValueError(
        "Not enough consecutive academic-session records "
        "for a chronological train/test split."
    )

# Handle missing numeric values and encode class labels.
preprocessor = ColumnTransformer([
    (
        "numeric",
        SimpleImputer(strategy="median"),
        numeric_features
    ),
    (
        "class",
        OneHotEncoder(handle_unknown="ignore"),
        categorical_features
    )
])

model = Pipeline([
    ("preprocessor", preprocessor),
    (
        "regressor",
        RandomForestRegressor(
            n_estimators=200,
            max_depth=12,
            min_samples_leaf=2,
            random_state=42,
            n_jobs=-1
        )
    )
])

# Train
model.fit(X_train, y_train)

# Evaluate on the latest target session
predictions = model.predict(X_test)

mae = mean_absolute_error(y_test, predictions)
rmse = np.sqrt(mean_squared_error(y_test, predictions))
r2 = r2_score(y_test, predictions)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
print("Test target session:", latest_target_year)
print(f"MAE:  {mae:.2f} percentage points")
print(f"RMSE: {rmse:.2f} percentage points")
print(f"R²:   {r2:.4f}")

# Save the complete preprocessing + model pipeline.
joblib.dump(model, "student_performance_model.pkl")
```
